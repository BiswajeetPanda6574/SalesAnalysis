# Exploratory Analysis using visuals

```python
sns.barplot(x="Category", y="Sales", data=df2, estimator=sum)
plt.title("Total Sales by Category")
plt.show()
```

```python
sns.barplot(x="Segment", y="Sales", data=df2, estimator=sum)
plt.title("Total Sales by Segment")
plt.show()
```

```python
sns.barplot(x="Sub-Category", y="Sales", data=df2, estimator=sum)
plt.title("Total Sales by Sub-CAtegory")
plt.xticks(rotation=45, ha="right")
plt.figure(figsize=(18,6))
plt.show()
```

```python
plt.figure(figsize=(18,10))  
sns.barplot(y="State", x="Sales", data=df)
plt.title("State wise Sales", fontsize=16)
plt.xlabel("Sales")
plt.ylabel("State")
plt.show()
```

```python
sns.barplot(x="Ship Mode", y="Sales", data=df2, estimator=sum, alpha=0.6)
plt.title("Total Sales via Shipping")
plt.show()
```

```python
sns.barplot(y = 'Region', x = 'Ship Mode', data = df2,
           hue = 'Segment',
           palette = 'inferno',
           errcolor = 'magenta',
           errwidth = 2)
```

```python
sns.barplot(x="Region", y="Sales", data=df2, estimator=sum)
plt.title("Total Sales by Region")
plt.show()
```

```python
sns.histplot(df2["Sales"], bins=30, kde=True)
```

```python
df2["Order Date"] = pd.to_datetime(df2["Order Date"], dayfirst=True)
df2["Ship Date"] = pd.to_datetime(df2["Ship Date"], dayfirst=True)
```

```python
df2["Order Date"] = pd.to_datetime(df2["Order Date"])
sales_over_time = df2.groupby("Order Date")["Sales"].sum().reset_index()
plt.figure(figsize=(12,6))
plt.plot(sales_over_time["Order Date"], sales_over_time["Sales"])
plt.title("Sales Trend Over Time")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df2["Sales"], bins=50, kde=True)
plt.xscale("log")
plt.title("Distribution of Sales (Log Scale)")
plt.xlabel("Sales (log scale)")
plt.ylabel("Frequency")
plt.show()
```

```python
bins = [2,3,4,5,6,7,8,9,10]
sns.distplot(df2['Postal Code'], bins = bins) # Custom bin edges for histogram
plt.xticks(bins) # Sets the x-axis ticks to match the bins
```

# INSIGHTS:

1. Technology category is the major consumer of the products which in return contributes the highest profit.
2. Among all the 4 regions(East, West, Central, South), West region has the highest and South region has the lowest sales overall.
3. The consumers are involed the most in purchase rather than corporate and home office.
4. Some products show high sales but negative profit.
5. Majority of the customers are preferring Standard class as their shipment mode and very few are preferring the order on the same day.
6. Sales trend shows seasonality/peak in certain months. But every year after Jan the sales gets decreased.7. 
7. Within the categories PHones are the most selled product.
8. Among the 48 States Wyoming is the majority buyer of the products while South Dakota has the minimum.
9. The highest sale in any month was in the year 2015 but the frequency of sales is highest in the year 2018.
